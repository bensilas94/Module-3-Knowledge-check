**Describe the two dominant metocean seasons in the Gulf of Guinea. How does the wave energy spectrum differ between dry season and wet season?**

Answer: The Two Dominant Metocean Seasons in the Gulf of Guinea.

The Gulf of Guinea, particularly offshore Nigeria (Usan, Bonga, Jubilee and surrounding fields), experiences two dominant metocean seasons:

1. Dry season (November–March)
2. Wet season (April–October)

DRY SEASON: The dry season is generally characterised by:
- Relatively calmer sea conditions.
- Lower local wind speeds over much of the region.
- Reduced rainfall and cloud cover.
- Greater influence of long-period Atlantic swell waves.
- More stable offshore operating conditions, although large swell events can still occur.

Wave characteristics:
- Lower contribution from locally generated wind waves.
- Swell waves travelling from the southern Atlantic can dominate.
- Wave energy is often concentrated at lower frequencies (longer wave periods).


WET SEASON: The wet season is generally characterised by:
- Increased rainfall and cloud cover.
- Stronger and more variable atmospheric and wind conditions.
- Greater weather-related sea-state variability.
- Increased contribution from locally generated wind waves.
- Potentially rougher sea conditions during energetic weather events.

Wave characteristics:
- Greater contribution from wind-generated waves.
- More energy can occur at higher frequencies (shorter wave periods).
- The wave spectrum may become broader, depending on the wind and swell conditions.



**How does the wave energy spectrum differ?**

The wet season spectrum shows higher total energy and bimodal distributions (combined swell and wind-sea), whereas the dry season spectrum exhibits lower energy and is heavily skewed toward longer-period swell components.

**QUESTION 2.**

Why is cyclical encoding (sine/cosine) necessary for month and hour features in ML models? Show the encoding for month = 9 (September).

Answer: Cyclical encoding is a feature engineering technique used to represent repeating time-based features, such as months of the year and hours of the day, so that ML models understand their circular nature

Why is it necessary?

Months and hours follow a repeating cycle.

For example:
- December (12) is immediately followed by January (1).
- Hour 23 is immediately followed by hour 0.
However, if we use their original numerical values, an ML model may interpret December (12) and January (1) as being far apart.

This is misleading because they are actually consecutive months.

Sine and cosine encoding solves this problem by converting the values into coordinates on a circle, preserving their cyclical relationship.

Encoding for month = 9 (September)

:$$month\_sin = \sin\left(\frac{2 \pi \times 9}{12}\right) = \sin\left(\frac{3\pi}{2}\right) = -1.0$$$$month\_cos = \cos\left(\frac{2 \pi \times 9}{12}\right) = \cos\left(\frac{3\pi}{2}\right) = 0.0$$

**QUESTION 3**
What probability distribution best fits significant wave height ($H_s$) data? Explain the physical justification and name the Python function to fit it

Answer:

The Weibull distribution is commonly used to model significant wave height (Hs) data in offshore and metocean engineering.

It is particularly useful for describing the statistical distribution of wave heights and estimating the probability of different sea states occurring.
1. Why is the Weibull distribution suitable?
The Weibull distribution is suitable because:
- Wave heights are non-negative (Hs cannot be negative).
- Wave heights are often positively skewed: lower and moderate waves occur more frequently than extremely high waves.
- It can represent different shapes of wave-height distributions by adjusting its shape and scale parameters.
- It is useful for estimating the probability of extreme wave conditions.
2. Physical justification
The Weibull distribution is physically reasonable because ocean waves are generated by a combination of wind, swell, and changing atmospheric conditions.

These processes produce:
- Many small-to-moderate wave heights.
- Fewer large wave heights.
- Very few extreme wave events.
The Weibull distribution can capture this positively skewed pattern.

```python
from scipy.stats import weibull_min

# Example significant wave height data
Hs = [0.8, 1.1, 1.3, 1.5, 1.8, 2.1, 2.5, 3.0]

# Fit Weibull distribution
shape, loc, scale = weibull_min.fit(Hs, floc=0)

print("Shape:", shape)
print("Location:", loc)
print("Scale:", scale)
```

**QUESTION 4.**

Explain the difference between the Block Maxima (GEV) and Peaks Over Threshold (GPD) methods for extreme value analysis. When would you prefer each?

Answer: Block Maxima (GEV): Divides the time series into non-overlapping blocks (typically annual) and extracts only the single maximum value per block, fitting a Generalised Extreme Value (GEV) distribution.

Peaks Over Threshold (POT): Selects all independent peak values that exceed a sufficiently high numerical threshold and fits a Generalised Pareto Distribution (GPD).

**When would you prefer each method?**

Prefer Block Maxima (GEV) when:
- You have a long historical dataset.
- You want to estimate annual maximum significant wave heights.
- You need annual extreme wave statistics for offshore structural design.
- You want to estimate 10-year, 50-year or 100-year return levels.


Prefer Peaks Over Threshold (GPD) when:
- You want to use more of the extreme observations.
- Your dataset contains multiple significant storm events.
- You want a more detailed analysis of extreme wave conditions.
- You have sufficient data to select a reliable threshold and identify independent events.

**QUESTION 6.**

Explain the five-step metocean-to-fatigue pipeline. Which step introduces the most uncertainty and why?


Answers:

The five-step pipeline
1. Step 1: Metocean data collection
   
   Collect environmental data such as:
   - Significant wave height (Hs)
   - Peak wave period (Tp)
   - Current speed and direction (Uc)
   - Wind speed and direction
   These describe the environmental conditions acting on the offshore asset.
2. Step 2: Feature engineering
   
   Convert raw metocean data into useful ML model inputs.
   
   Examples:
   - Rolling mean and maximum of Hs
   - Wave energy
   - Cumulative wave exposure
   - Hours since calm conditions
   - Seasonal and tidal features
   This captures not just current conditions but also their history.
3. Step 3: Structural response calculation
   
   Estimate how the asset responds to environmental loading.
   
   Examples:
   - Riser displacement
   - Vortex-induced vibration (VIV)
   - Stress range (Δσ)
   - Stress cycles
   Waves and currents generate loads that can cause structural vibration and cyclic stress.
4. Step 4: Fatigue damage calculation
   
   Convert stress cycles into fatigue damage using an S–N curve and Miner's Rule.
   
   S–N relationship:
   \[
   \log N=\log \bar a-m\log\Delta\sigma
   \]
   
   Miner's Rule:
   \[
   D=\sum_i\frac{n_i}{N_i}
   \]
   
   Where:
   - \(n_i\) = actual number of cycles at a stress range
   - \(N_i\) = cycles to failure at that stress range
   - \(D\) = accumulated fatigue damage
5. Step 5: Prediction and asset integrity decision
   
   Use calculated fatigue damage to train and evaluate ML models, such as Random Forest and XGBoost.
   
   The model can help:
   - Predict future fatigue damage
   - Identify high-risk assets
   - Estimate remaining useful life (RUL)
   - Support inspection and maintenance planning


**Which step introduces the most uncertainty and why?**

Main uncertainty: Step 3 – Structural response calculation
Structural response calculation often introduces substantial uncertainty because converting environmental conditions into actual stress and vibration experienced by an offshore asset is complex.

Why?
1. Complex hydrodynamics: Wave and current interactions are difficult to predict accurately.
2. Asset characteristics: Riser geometry, material properties, support conditions and tension influence structural response.
3. Vortex-induced vibration (VIV): Current flowing around a riser can generate alternating forces and vibrations.
4. Operating conditions: Changes in pressure, temperature, flow rate and slugging can influence the structural response.
5. Modelling assumptions: Numerical models simplify real offshore conditions, which can introduce errors.
For example, two risers exposed to the same current speed may experience different stress ranges because their geometry, tension and boundary conditions differ.

Important distinction: Uncertainty can also enter at other stages. Poor-quality metocean measurements affect Step 1, while incorrect fatigue properties or S–N curves affect Step 4. The largest uncertainty depends on the asset, data quality and modelling approach.
Pipeline summary


Step	Main activity	Output
1	Metocean data collection	Hs, Tp, Uc
2	Feature engineering	ML-ready features
3	Structural response	Stress ranges and cycles
4	Fatigue calculation	Fatigue damage
5	ML prediction	Future damage and integrity decisions

**QUESTION 10.**

The JONSWAP peak enhancement factor ($\gamma$) is typically 3.3 for the North Sea but 2.0–2.5 for the Gulf of Guinea. What physical characteristic of the wave climate explains this difference?

Answer:

The difference is explained by wave-generation mechanics and fetch geometry.

The North Sea is characterized by shorter fetches and steep, sharply peaked wind-sea waves driven by severe storms, yielding a higher peak enhancement factor ($\gamma = 3.3$). Conversely, the Gulf of Guinea is dominated by long-period, highly dispersed South Atlantic swells generated by distant meteorological events, resulting in a broader, less sharply peaked energy spectrum that corresponds to a lower $\gamma$ ($2.0–2.5$).
